# 01 · Explore labels
ToxCast/Tox21 hit-calls and endpoint annotations from the authors' repo; native MT / LDH / cell-count hits from the SI table.

```python
import sys; sys.path.insert(0, '../src')
import pandas as pd, numpy as np
pd.set_option('display.width', 200)
from cpsa import RESULTS, PROCESSED
from cpsa.data.load_labels import load_all_labels
labels, endpoints = load_all_labels()
print(labels.shape)
endpoints.category.value_counts()
```

```python
n1 = (labels==1).sum(); n0 = (labels==0).sum()
summary = pd.DataFrame({'n_active': n1, 'n_inactive': n0}).join(endpoints)
summary['powered_15'] = (summary.n_active>=15)&(summary.n_inactive>=15)
print(summary.groupby('category').powered_15.agg(['sum','count']))
summary.n_active.describe()
```
